#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
march_madness_summary <- march_madness |> 
  summarize(
    mean_year = mean(year, na.rm = TRUE),
    min_year = min(year, na.rm = TRUE),
    max_year = max(year, na.rm = TRUE),

    mean_seed = mean(seed, na.rm = TRUE),
    min_seed = min(seed, na.rm = TRUE),
    max_seed = max(seed, na.rm = TRUE),

    mean_tourney_wins = mean(tourney_wins, na.rm = TRUE),
    min_tourney_wins = min(tourney_wins, na.rm = TRUE),
    max_tourney_wins = max(tourney_wins, na.rm = TRUE))
      
march_madness_summary
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
march_madness_missing_data <- march_madness |> 
  summarize(across(c(school, year, seed, bid, tourney_wins), ~ sum(is.na(.))))

print(march_madness_missing_data)
```

```r
tourney_wins_graph <- ggplot(march_madness, aes(x = tourney_wins)) + 
  geom_histogram(bins = 12, fill = "light blue", color = "white") +
  labs(
    title = "Distribution of Tournament Wins",
    x = "Tournament Wins",
    y = "Number of Teams") +
  theme_minimal()

tourney_wins_graph
```

```r
march_madness_summary <- march_madness |>
  count(bid) |>
  mutate(percentage = n / sum(n))

bid_type_graph <- ggplot(march_madness_summary, aes(x = bid, y = percentage)) +
  geom_col(fill = "light coral") +
  labs(
    title = "Percentage of Teams by Bid Type",
    x = "Bid Type",
    y = "Percentage of Teams")

bid_type_graph
```

Visualization 1: This scatterplot shows the relationship between a team's seed and the number of tournament games a team wins in March Madness.
```r
### EDIT THIS CODE CELL FOR VISUALIZATION 1###

seed_wins <- ggplot(march_madness, aes(x = seed, y = tourney_wins)) +
  geom_point(alpha = 0.5) +
  labs(
    title = "Relationship Between Team Seed and Tournament Wins",
    x = "Team's Seed",
    y = "Number of Tournament Wins"
  )

seed_wins
```

Visualization 2: This bar plot shows the relationship between a team's bid type and their total tournament wins in March Madness. Each bar represents two bid types, auto and at-large, on the x-axis while the total number of tournament wins is represented on the y-axis. This is relevant to the analysis because it shows that teams with an at-large bid have significantly more tournament wins compared to teams with an auto bid. This indicates that bid type is another predictor for determining the number of tournament wins for a team. 

```r
bid_wins <- ggplot(march_madness, aes(x = bid, weight = tourney_wins)) +
  geom_bar(fill = "light blue") +
  labs(
    title = "Total Tournament Wins by Bid Type",
    x = "Bid Type",
    y = "Total Number of Tournament Wins") +
  theme_minimal()

bid_wins
```
